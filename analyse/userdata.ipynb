#### Узнаем максимальный возможный индекс пользователя

```python
left = 1
right = 2_000_000_000
while right - left > 1:
    mid = (right + left) // 2
    user = get_userdata(mid)
    print(f"[{left}..{right}]", right - left)
    if user is None:
        right = mid
    else:
        left = mid

MAX_ACCOUNT_ID = right
```

#### Узнаем минимальный индекс аккаунта

```python
left = 1
right = MAX_ACCOUNT_ID
while right - left > 1:
    mid = (right + left) // 2
    user = get_userdata(mid)
    assert user is not None
    print(f"[{left}..{right}]", right - left)
    if user.createdAt.year == 2025 and user.createdAt.month >= 4: # Account created since april 2025
        right = mid
    else:
        left = mid

MIN_ACCOUNT_ID = right
```

```python
print(MIN_ACCOUNT_ID, MAX_ACCOUNT_ID)
print("Total accounts created since 01.04.2025:", MAX_ACCOUNT_ID - MIN_ACCOUNT_ID + 1)
```
